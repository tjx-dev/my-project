## 变量与数据类型

```python
#include <stdio.h>
#include <limits.h>
#include <float.h>

int main() {
    int a = 42;
    float b = 3.14f;
    double c = 2.718281828;
    char d = 'A';
    
    printf("int: %d (大小: %zu 字节, 范围: %d ~ %d)\n", a, sizeof(int), INT_MIN, INT_MAX);
    printf("float: %.2f (大小: %zu 字节)\n", b, sizeof(float));
    printf("double: %.9f (大小: %zu 字节)\n", c, sizeof(double));
    printf("char: %c (大小: %zu 字节)\n", d, sizeof(char));
    return 0;
}
```

## 条件判断与循环

```python
#include <stdio.h>

int main() {
    // if-else 条件判断
    int score = 85;
    printf("分数: %d → ", score);
    if (score >= 90) {
        printf("优秀\n");
    } else if (score >= 60) {
        printf("及格\n");
    } else {
        printf("不及格\n");
    }
    
    // for 循环 - 打印乘法表
    printf("\n9×9乘法表:\n");
    for (int i = 1; i <= 9; i++) {
        for (int j = 1; j <= i; j++) {
            printf("%d×%d=%-3d", j, i, i * j);
        }
        printf("\n");
    }
    return 0;
}
```

## 数组操作

```python
#include <stdio.h>

int main() {
    int arr[] = {5, 2, 8, 1, 9, 3};
    int n = sizeof(arr) / sizeof(arr[0]);
    
    printf("原始数组: ");
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");
    
    // 冒泡排序
    for (int i = 0; i < n - 1; i++) {
        for (int j = 0; j < n - 1 - i; j++) {
            if (arr[j] > arr[j + 1]) {
                int tmp = arr[j];
                arr[j] = arr[j + 1];
                arr[j + 1] = tmp;
            }
        }
    }
    
    printf("排序后:   ");
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");
    
    // 求和与平均值
    int sum = 0;
    for (int i = 0; i < n; i++) sum += arr[i];
    printf("总和: %d, 平均值: %.2f\n", sum, (float)sum / n);
    return 0;
}
```

## 指针与函数

```python
#include <stdio.h>

// 通过指针交换两个数
void swap(int *a, int *b) {
    int tmp = *a;
    *a = *b;
    *b = tmp;
}

// 递归计算阶乘
int factorial(int n) {
    if (n <= 1) return 1;
    return n * factorial(n - 1);
}

int main() {
    // 指针与交换
    int x = 10, y = 20;
    printf("交换前: x = %d, y = %d\n", x, y);
    swap(&x, &y);
    printf("交换后: x = %d, y = %d\n", x, y);
    
    // 递归阶乘
    int n = 6;
    printf("\n%d! = %d\n", n, factorial(n));
    
    // 指针与数组
    int nums[] = {10, 20, 30, 40, 50};
    int *p = nums;
    printf("\n通过指针遍历数组: ");
    for (int i = 0; i < 5; i++) {
        printf("%d ", *(p + i));
    }
    printf("\n");
    return 0;
}
```

## 结构体

```python
#include <stdio.h>
#include <string.h>

struct Student {
    char name[20];
    int age;
    float score;
};

int main() {
    struct Student s1;
    strcpy(s1.name, "张三");
    s1.age = 20;
    s1.score = 92.5f;
    
    struct Student s2 = {"李四", 21, 88.0f};
    
    printf("学生信息:\n");
    printf("  %-8s %s  %s\n", "姓名", "年龄", "分数");
    printf("  %-8s %-4d %.1f\n", s1.name, s1.age, s1.score);
    printf("  %-8s %-4d %.1f\n", s2.name, s2.age, s2.score);
    return 0;
}
```

## 动态内存分配

```python
#include <stdio.h>
#include <stdlib.h>

int main() {
    int n = 5;
    
    // malloc 分配内存
    int *arr = (int *)malloc(n * sizeof(int));
    if (arr == NULL) {
        printf("内存分配失败!\n");
        return 1;
    }
    
    // 初始化
    for (int i = 0; i < n; i++) {
        arr[i] = (i + 1) * (i + 1);  // 平方数
    }
    
    printf("动态数组 (平方数): ");
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");
    
    // 释放内存
    free(arr);
    printf("内存已释放\n");
    return 0;
}
```

## 文件操作

```python
#include <stdio.h>

int main() {
    // 写入文件
    FILE *fp = fopen("test_c_notebook.txt", "w");
    if (fp == NULL) {
        printf("无法打开文件!\n");
        return 1;
    }
    
    fprintf(fp, "这是从 C Jupyter Notebook 写入的文本\n");
    fprintf(fp, "C 语言文件操作示例\n");
    fclose(fp);
    printf("文件写入成功: test_c_notebook.txt\n");
    
    // 读取文件
    fp = fopen("test_c_notebook.txt", "r");
    if (fp == NULL) {
        printf("无法读取文件!\n");
        return 1;
    }
    
    printf("\n读取文件内容:\n");
    printf("---\n");
    char buffer[256];
    while (fgets(buffer, sizeof(buffer), fp) != NULL) {
        printf("%s", buffer);
    }
    printf("---\n");
    fclose(fp);
    return 0;
}
```
