```python
#封装  把属性和方法封装一起，仅给别人去访问 
class Bankaccount:
    def __init__(self,owner:str,balance:float=0):
        self.owner=owner
        self.__balance=balance   # 单下划线约定为“受保护”，不要外部直接改

# getter：像访问属性一样调用，但内部可以加逻辑
    @property
    def balance(self):
        return self.__balance
# setter：控制写入，防止负数
    @balance.setter
    def balance(self,value):
        if value<0:
            raise ValueError("余额不能为负数")
        self.__balance=value
        #取钱
    def withdraw(self,amount):
        if amount>self.__balance:
            raise ValueError("余额不足")
        self.__balance-=amount
        return amount
user1=Bankaccount("Alice",1000)
user1.withdraw(120) # 取款
print(user1.balance)  # 访问余额
user1.balance=500 # 修改余额
user1.withdraw(200) # 再次取款
print(user1.balance)  # 再次访问余额
```

```python
#继承和方法重写
class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        return "sound of animal"
class Dog(Animal):
    # 子类自己的初始化
    def __init__(self,name:str,breed:str):
        super().__init__(name)  # 必须显式调父类构造
        self.breed=breed
        # 完全重写父类方法
    def speak(self):
        return f"{self.name} says Woof!" 
         # 也可以扩展父类方法
    def info(self):
        return f"{self.name} is a {self.breed} breed."
class Cat(Animal):
    def speak(self):
        return f"{self.name} says Meow!"
    
animal=[Dog("Buddy","Labrador"),Cat("Whiskers")]
for a in animal:
    print(a.speak())
print(Dog("Buddy","Labrador").info())
```

```python
#多态与抽象基类(ABC)
from abc import ABC,abstractmethod
class Notifier(ABC):
    @abstractmethod
    def send(self,message:str)->bool:
        """所有通知方式必须实现 send 方法"""
        pass
class EmailNotifier(Notifier):
    def send(self,message:str)->bool:
        print(f"Sending email: {message}")
        return True
class SMSNotifier(Notifier):
    def send(self,message:str)->bool:
        print(f"Sending SMS: {message}")
        return True
#不能实例化抽象类
# notifier=Notifier()  # 会报错
def alter(Notifier,message:str):
    Notifier.send(message)
alter(EmailNotifier(),"Hello via Email!")
alter(SMSNotifier(),"Hello via SMS!")
```

```python
#魔法方法：让对象行为像内置类型
class Book:
    def __init__(self,title,author,pages):
        self.title=title
        self.author=author
        self.pages=pages
    def __str__(self):
        """print() 和 str() 时调用，给用户看的友好信息"""
        return f"《{self.title}》 by {self.author}, {self.pages} pages"
    def  __repr__(self):
        """repr() 时调用，给开发者看的详细信息"""
        return f"Book('{self.title}', '{self.author}', {self.pages})"
    def __eq__(self,other):
        """== 比较时调用，比较内容而不是地址"""
        if  not isinstance(other,Book):
            return False
        return (self.title,self.author,self.pages)==(other.title,other.author,other.pages)
    def __len__(self):
        """len() 时调用，返回书的页数"""
        return self.pages
book1=Book("1984","George Orwell",328)
book2=Book("1984","George Orwell",327)
print(book1)  # 调用 __str__
print(repr(book1))  # 调用 __repr__
print(book1 == book2)  # 调用 __eq__
print(len(book1))  # 调用 __len__
```
